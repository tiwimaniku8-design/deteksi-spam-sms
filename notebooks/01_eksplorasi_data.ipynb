```python
import pandas as pd

url = "https://raw.githubusercontent.com/tiwimaniku8-design/deteksi-spam-sms/main/data/raw/SMSSpamCollection"
df = pd.read_csv(url, sep="\t", header=None, names=["label", "message"])

print(df.shape)
print(df["label"].value_counts())
print(df.head())
```

```python
# Cek pesan kosong dan duplikat
print("Pesan kosong:", df["message"].isna().sum())
print("Pesan duplikat:", df.duplicated().sum())

# Panjang pesan per kelas
df["panjang"] = df["message"].str.len()
print(df.groupby("label")["panjang"].describe())

# Contoh spam untuk bahan presentasi
print("\nContoh spam:")
for msg in df[df["label"] == "spam"]["message"].head(5):
    print("-", msg)
```

```python
dup = df[df.duplicated(keep=False)]
print(dup["label"].value_counts())
```

```python
df_unik = df.drop_duplicates()
print(df_unik["label"].value_counts())
print("Persentase spam:", round(df_unik["label"].eq("spam").mean() * 100, 2), "%")
```
